```python
import matplotlib.pyplot as plt
```

```python
# Data
languages = ['Python', 'Rust']
ram = [18, 2.2]
time = [2084, 167]
```

```python
# Apply a clean aesthetic style
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4.5), dpi=300)

# Colors: Python blue vs. Rust vibrant orange/rust tone
colors = ['#4B8BBE', '#DEA584']

# --- RAM Chart ---
bars1 = ax1.bar(languages, ram, color=colors, width=0.55, edgecolor='#333333', linewidth=0.8)
ax1.set_title('Memory Usage (RAM)', fontsize=12, fontweight='bold', pad=12)
ax1.set_ylabel('Megabytes (MB)', fontsize=10, fontweight='semibold')
ax1.set_ylim(0, max(ram) * 1.25)  # Leave room for text labels
ax1.grid(axis='y', linestyle='--', alpha=0.7)

# Add value labels on top of bars
for bar in bars1:
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height + 0.6,
             f'{height} MB',
             ha='center', va='bottom', fontsize=10, fontweight='bold')

# --- Time Chart ---
bars2 = ax2.bar(languages, time, color=colors, width=0.55, edgecolor='#333333', linewidth=0.8)
ax2.set_title('Execution Time', fontsize=12, fontweight='bold', pad=12)
ax2.set_ylabel('Nanoseconds (ns)', fontsize=10, fontweight='semibold')
ax2.set_ylim(0, max(time) * 1.25)
ax2.grid(axis='y', linestyle='--', alpha=0.7)

# Add value labels on top of bars
for bar in bars2:
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height + 90,
             f'{height:,} ns',
             ha='center', va='bottom', fontsize=10, fontweight='bold')

# General layout cleanup
fig.suptitle('Performance Comparison: Python Map vs. Rust Bitmask Approach', fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()

# Save high-resolution image for docs/READMEs
plt.savefig('performance_comparison.png', bbox_inches='tight')
plt.show()
```
